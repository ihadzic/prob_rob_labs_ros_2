```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
zoxo=[]
zcxo=[]
zoxc=[]
zcxc=[]

f = open("opened.txt", 'r')
set = f.readlines()

for data in set:
    data = float(data)
    if (data <= 321.55):
        zoxo.append(data)
    else:
        zcxo.append(data)


f = open("closed.txt", 'r')
set = f.readlines()

for data in set:
    data = float(data)
    if (data >= 321.55):
        zcxc.append(data)
    else:
        zoxc.append(data)

```

```python
p_zoxo = len(zoxo) / (len(zoxo) + len(zcxo))
p_zcxo = len(zcxo) / (len(zoxo) + len(zcxo))
p_zoxc = len(zoxc) / (len(zoxc) + len(zcxc))
p_zcxc = len(zcxc) / (len(zoxc) + len(zcxc))

print(f"P(z=open | x=open) = {p_zoxo:.4f}")
print(f"P(z=closed | x=open) = {p_zcxo:.4f}")
print(f"P(z=open | x=closed) = {p_zoxc:.4f}")
print(f"P(z=closed | x=closed) = {p_zcxc:.4f}")
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(np.arange(len(zoxo)), zoxo, label='z=open, x=open', color='green', s=10)
plt.scatter(np.arange(len(zcxo)), zcxo, label='z=closed, x=open', color='red', s=10)
plt.scatter(np.arange(len(zoxc)), zoxc, label='z=open, x=closed', color='orange', s=10)
plt.scatter(np.arange(len(zcxc)), zcxc, label='z=closed, x=closed', color='blue', s=10)

plt.axhline(y=321.55, color='black', linestyle='--', label='Threshold')

plt.xlabel('Sample Index')
plt.ylabel('Feature Mean')
plt.title('Door State Classification')
plt.legend()
plt.grid(True)
plt.show()
```
